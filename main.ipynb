# Kidney Disease Screening: Descriptive Statistics

This analysis examines synthetic patients with eGFR ≤ 65. Height and weight are assumed to be inches and pounds; verify source units before interpreting BMI.

```python
import pandas as pd
import matplotlib.pyplot as plt
from ckd_screening import load_patients

# Retrieve and harmonize the patient datasets
patients = load_patients()
patients.head()
```

## Descriptive statistics
Summarize the distribution of patient age, height, and weight.

```python
patients[['age', 'height', 'weight']].describe().round(2)
```

## Average eGFR by BMI category
Use standard adult BMI cutoffs.

```python
# Categorize BMI and calculate the mean eGFR in each group
labels = ['Underweight', 'Healthy weight', 'Overweight', 'Obesity']
patients['BMI_category'] = pd.cut(patients['BMI'], bins=[0, 18.5, 25, 30, float('inf')], labels=labels, right=False)
mean_egfr = patients.groupby('BMI_category', observed=False)['eGFR'].mean()
mean_egfr.plot(kind='bar', rot=0, figsize=(8, 4))
plt.title('Average eGFR by BMI Category')
plt.xlabel('BMI category')
plt.ylabel('Average eGFR (mL/min/1.73 m²)')
plt.tight_layout()
plt.show()
```

## Age and eGFR
Examine the relationship between patient age and eGFR.

```python
plt.figure(figsize=(7, 4))
plt.scatter(patients['age'], patients['eGFR'], alpha=0.6)
plt.xlabel('Age (years)')
plt.ylabel('eGFR (mL/min/1.73 m²)')
plt.title('Patient Age vs. eGFR')
plt.tight_layout()
plt.show()
```

## Patient sex
Show the relative proportions of patients by recorded sex.

```python
patients['sex'].value_counts().plot(kind='pie', autopct='%1.1f%%', ylabel='')
plt.title('Patient Sex Distribution')
plt.tight_layout()
plt.show()
```
